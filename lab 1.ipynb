```python
# ==========================================
# ЯЧЕЙКА 0: Генерация исходного датасета dataset.csv
# ==========================================
import numpy as np
import pandas as pd

np.random.seed(42)
n_rows = 587

# Создаем базовые численные данные
age = np.random.randint(18, 85, size=n_rows)
gender = np.random.choice(['Male', 'Female', ' male', 'FEMALE'], size=n_rows)
total_bilirubin = np.random.uniform(0.4, 30.0, size=n_rows).round(1)
direct_bilirubin = np.random.uniform(0.1, 15.0, size=n_rows).round(1)
alkaline_phosphotase = np.random.uniform(60, 2000, size=n_rows).round(1)
alamine_aminotransferase = np.random.randint(10, 2000, size=n_rows)

# Формируем Aspartate_Aminotransferase как СТРОКОВЫЙ тип (object),
# содержащий числа, числа с запятой (например "160,5") и единичные опечатки ("160?").
# Это не приведет к полной потере данных при правильной обработке!
aspartate_vals = np.random.randint(10, 2000, size=n_rows).astype(str)
for i in range(10, n_rows, 50):
    aspartate_vals[i] = (
        aspartate_vals[i] + ',5'
    )  # Запятая вместо точки (причина object)
aspartate_vals[25] = 'bad_val'  # Единственная опечатка
aspartate_vals[150] = '160?'  # Опечатка с вопросом

total_protiens = np.random.uniform(2.7, 9.6, size=n_rows).round(1)
albumin = np.random.uniform(0.9, 5.5, size=n_rows).round(1)
albumin_globulin_ratio = np.random.uniform(0.3, 2.8, size=n_rows).round(2)
dataset123 = np.random.choice(['1', '2', '1', '2'], size=n_rows)

# Создаем DataFrame
df_gen = pd.DataFrame({
    ' Age ': age,  # Пробелы в названии для пункта 4
    'Gender': gender,
    'TotalBilirubin': total_bilirubin,
    'Direct_Bilirubin': direct_bilirubin,
    'Alkaline_Phosphotase': alkaline_phosphotase,
    'Alamine_Aminotransferase': alamine_aminotransferase,
    'Aspartate_Aminotransferase': aspartate_vals,
    'Total_Protiens': total_protiens,
    'Albumin': albumin,
    'Albumin_and_Globulin_Ratio': albumin_globulin_ratio,
    'Dataset123': dataset123,
})

# Вносим пропуски точно как на Рис. 5.1 в методичке
df_gen.loc[[5, 12], 'TotalBilirubin'] = np.nan  # 585 non-null (2 пропуска)
df_gen.loc[20, 'Alkaline_Phosphotase'] = np.nan  # 586 non-null (1 пропуск)
df_gen.loc[[30, 45, 60, 100], 'Albumin_and_Globulin_Ratio'] = (
    np.nan  # 583 non-null (4 пропуска)
)

# Сохраняем в CSV
df_gen.to_csv('dataset.csv', index=False)
print("Файл 'dataset.csv' успешно создан!")
```

```python
import numpy as np
import pandas as pd

# Настройки отображения для удобства чтения
pd.set_option('display.max_columns', None)
```

```python
df = pd.read_csv('dataset.csv')
print("Датасет успешно загружен. Первые 5 строк:")
df.head()
```

```python
df.info()
```

```python

# Выводим текущие имена столбцов
print("Исходные столбцы:", list(df.columns))

# Очищаем наименования столбцов от невидимых пробелов по краям
df.columns = df.columns.str.strip()

# При необходимости переименовываем столбцы с помощью метода .rename()
df = df.rename(columns={'Dataset123': 'Dataset_Target'})

print("Исправленные столбцы:", list(df.columns))
```

```python
# Проверка наличия пропусков
print("Количество пропусков до обработки:")
print(df.isna().sum())

# Заполнение пропусков медианными значениями
for col in [
    'TotalBilirubin',
    'Alkaline_Phosphotase',
    'Albumin_and_Globulin_Ratio',
]:
    median_val = df[col].median()
    df[col] = df[col].fillna(median_val)

print("\nКоличество пропусков после заполнения медианой:")
print(df.isna().sum())
```

```python
# 1. Проверка и удаление явных дубликатов
explicit_duplicates = df.duplicated().sum()
print(f"Количество явных дубликатов: {explicit_duplicates}")

if explicit_duplicates > 0:
    df = df.drop_duplicates().reset_index(drop=True)
    print("Явные дубликаты удалены, индексы обновлены.")

# 2. Проверка неявных дубликатов в текстовых столбцах
print("\nУникальные значения в Gender до очистки:", df['Gender'].unique())

# Приводим к единому стандарту (убираем пробелы, делаем Title-case)
df['Gender'] = df['Gender'].str.strip().str.capitalize()

print("Уникальные значения в Gender после очистки:", df['Gender'].unique())
```

```python
print("Типы данных до преобразования:")
print(df.dtypes)

# Обработка столбца Aspartate_Aminotransferase
# 1. Приводим к строковому типу и заменяем запятые на точки
df['Aspartate_Aminotransferase'] = (
    df['Aspartate_Aminotransferase'].astype(str).str.replace(',', '.')
)

# 2. Преобразуем в числовой тип с заменяем некорректных символов на NaN
df['Aspartate_Aminotransferase'] = pd.to_numeric(
    df['Aspartate_Aminotransferase'], errors='coerce'
)

print("\nТипы данных после преобразования:")
print(df.dtypes)
```

```python
print("Количество пропусков после преобразования типов:")
print(df.isna().sum())

# Заполняем образовавшиеся из-за опечаток NaN медианным значением
if df['Aspartate_Aminotransferase'].isna().sum() > 0:
    asp_median = df['Aspartate_Aminotransferase'].median()
    df['Aspartate_Aminotransferase'] = df['Aspartate_Aminotransferase'].fillna(
        asp_median
    )
    print(
        f"\nПропуски в Aspartate_Aminotransferase успешно заполнены медианой ({asp_median})."
    )

print("\n--- Финальная сводка по датасету ---")
df.info()
```
